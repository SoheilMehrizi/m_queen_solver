```python
from datetime import datetime as time
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def promising(board, row, col, n):
    for i in range(col):
        if board[row][i] == 1:
            return False
    for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
        if board[i][j] == 1:
            return False
    for i, j in zip(range(row, n, 1), range(col, -1, -1)):
        if board[i][j] == 1:
            return False
    return True

def solve_queen(board, col, n, solutions):
    if col == n:
        solutions.append([board[i].index(1) + 1 for i in range(n)])
        return
    for i in range(n):
        if promising(board, i, col, n):
            board[i][col] = 1
            solve_queen(board, col + 1, n, solutions)
            board[i][col] = 0

def m_queen(m):
    board = [[0 for _ in range(m)] for _ in range(m)]
    solutions = []
    solve_queen(board, 0, m, solutions)
    return solutions

```

# Visualization

```python
def plot_queen_solution(solution, board_size):
    board = [[0]*board_size for _ in range(board_size)]
    for col, row in enumerate(solution):
        board[row - 1][col] = 1

    plt.figure(figsize=(board_size, board_size))
    sns.heatmap(board, cbar=False, linewidths=1, linecolor='black',
                square=True, cmap='Blues', annot=True, fmt='d',
                xticklabels=[f'Col {i+1}' for i in range(board_size)],
                yticklabels=[f'Row {i+1}' for i in range(board_size)])
    
    plt.title("M-Queens Solution", fontsize=16)
    plt.yticks(rotation=0)
    plt.xticks(rotation=0)
    plt.show()
```
